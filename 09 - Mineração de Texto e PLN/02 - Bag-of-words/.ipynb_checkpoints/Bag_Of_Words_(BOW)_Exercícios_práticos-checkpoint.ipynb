#Questão 1
Complete a função a seguir para treinar um vetorizador com TF com IDF e retornar a matriz com a frequência dos termos, seguindo a definição da função a seguir. Nessas questões com o Code Runner, apenas utilize as classes e objetos passados como parâmetro para a função passada.

Apenas altere o código dentro da função e leia os comentários.

```python
def get_tf_idf (texts, TfidVectorizer): # Removed the second parameter as it's not needed
  # Instancia o vetorizador - configure o parâmetro de IDF (veja a documentação)
  #  - https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.TfidfVectorizer.html
  vectorizer = TfidfVectorizer(use_idf=True) # Use the imported class directly

  # Treina o vetorizador com os textos
  vectorizer.fit(texts)

  # Transforma e retorna a matriz de frequencia dos termos
  result = vectorizer.transform(texts) # Transform the input texts

  return result


texts = ['Mineração de Dados foi interessante.',
         'Perceptron é o começo de Redes Neurais.']

print(get_tf_idf(texts, None))
```

```python
texts = [
  "Rapadura é doce mas não é mole.",
  "Coca-cola zero não é doce.",
  "Cerveja não é água.",
  "Sorvete de maracujá acalma.",
  "Caldo de cana é doce."
]

def get_tfs (texts):
  # Instanciar o vetorizador sem o IDF
  vectorizer = TfidfVectorizer(use_idf=False)

  # Treina o vetorizador com os textos
  vectorizer.fit(texts)

  # Retorna a matriz de frequencia dos termos
  return vectorizer.transform(texts)

print(get_tfs(texts)[1].toarray().tolist())
```
